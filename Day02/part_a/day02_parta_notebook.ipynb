```python
import os

filename = "../input.txt"
with open(filename) as opened_file:
    line = opened_file.readline()

id_list = line.split(",")
inv_id_sum = 0

for id in id_list:
    first, last = id.split("-")
    for i in range(int(first), int(last)+1, 1):
        if len(str(i)) % 2 == 0:
            str_i = str(i)
            id1 = str_i[:len(str_i)//2]
            id2 = str_i[len(str_i)//2:]
            if id1 == id2:
                inv_id_sum += i
print(inv_id_sum)
```
