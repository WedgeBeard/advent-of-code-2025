```python
import os

filename = "../testinput.txt"
with open(filename) as opened_file:
    line = opened_file.readline()

id_list = line.split(",")
inv_id_sum = 0

for id in id_list:
    first, last = id.split("-")
    for i in range(int(first), int(last)+1, 1):
        print(f"Checking i: {i}")
        if len(str(i)) % 2 == 0:
            str_i = str(i)
            id1 = str_i[:len(str_i)//2]
            id2 = str_i[len(str_i)//2:]
            if id1 == id2:
                print(f"FOUND dub: {id1} {id2}")
                inv_id_sum += i
        if len(str(i)) % 3 == 0:
            id1 = str_i[:len(str_i)//3]
            id2 = str_i[len(str_i)//3:(len(str_i)//3)*2]
            id3 = str_i[(len(str_i)//3)*2:]
            print(f"compare trip: id1:{id1} id2:{id2} id3:{id3}")
            if id1 == id2 == id3:
                print(f"FOUND trip: {id1} {id2} {id3}")
                inv_id_sum += i
        if len(str(i)) % 5 == 0:
            id1 = str_i[:len(str_i)//5]
            id2 = str_i[len(str_i)//5:(len(str_i)//5)*2]
            id3 = str_i[(len(str_i)//5)*2:(len(str_i)//5)*3]
            id4 = str_i[(len(str_i)//5)*3:(len(str_i)//5)*4]
            id5 = str_i[(len(str_i)//5)*4:]
            if id1 == id2 == id3 == id4 == id5:
                print(f"FOUND quint: {id1} {id2} {id3} {id4} {id5}")
                inv_id_sum += i

print(inv_id_sum)


#longest is 10 digits long
```
