```python
import os
pos = 50
zero_times = 0
max = 100

filename = "../input.txt"
with open(filename) as opened_file:
    lines = opened_file.readlines()

for line in lines:
    if(line[0] == 'L'):
        val = int(line.strip('L'))
        pos = (pos - val) % max
    else:
        val = int(line.strip('R'))
        pos = (pos + val) % max
    if(pos == 0):
        zero_times += 1

print("Times 0 was landed on:", zero_times)
```
