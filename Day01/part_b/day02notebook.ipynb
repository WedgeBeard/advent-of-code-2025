```python
import os
pos = 50
zero_times = 0
max = 100

filename = "../testinput.txt"
with open(filename) as opened_file:
    lines = opened_file.readlines()

for line in lines:
    if(line[0] == 'L'):
        val = int(line.strip('L'))
        zero_times += int(val/max)
        pos = (pos - val) % max
        if pos - val < 0:
            zero_times += 1
    else:
        val = int(line.strip('R'))
        zero_times += int(val/max)
        pos = (pos + val) % max
        if pos + val > max:
            zero_times += 1
    
print("Times 0 was passed through:", zero_times)
#7045 too high
```
